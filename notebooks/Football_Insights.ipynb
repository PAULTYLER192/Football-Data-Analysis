# Football Insights

Module 4 visualizes the football results dataset with reusable match logic and summary charts.

```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Load the raw results data from the first path that exists so the notebook works in VS Code and Jupyter.
data_path_candidates = [
    Path("data/raw/results.csv"),
    Path("../data/raw/results.csv"),
    Path("../../data/raw/results.csv"),
]

for candidate in data_path_candidates:
    if candidate.exists():
        data_path = candidate
        break
else:
    raise FileNotFoundError("Could not locate data/raw/results.csv from the notebook environment.")

df = pd.read_csv(data_path)

# Parse dates and recreate the Module 2 and Module 3 logic inside the notebook.
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df["total_goals"] = df["home_score"] + df["away_score"]

def match_result(row):
    if row["home_score"] > row["away_score"]:
        return "Home Win"
    if row["home_score"] < row["away_score"]:
        return "Away Win"
    return "Draw"

df["result"] = df.apply(match_result, axis=1)

# The notebook now has the reusable columns needed for all remaining charts.
```

## Section 2: Goal Distribution

The histogram highlights how often low-scoring matches occur compared with more open, high-scoring games.

```python
plt.figure(figsize=(10, 6))
plt.hist(df["total_goals"], bins=15, color="#2a9d8f", edgecolor="black")
plt.title("Distribution of Goals Per Match")
plt.xlabel("Total Goals")
plt.ylabel("Number of Matches")
plt.tight_layout()
plt.show()
```

## Section 3: Match Outcomes

This bar chart compares how often matches end as home wins, away wins, or draws.

```python
outcome_counts = df["result"].value_counts().reindex(["Home Win", "Away Win", "Draw"])

plt.figure(figsize=(8, 6))
sns.barplot(x=outcome_counts.index, y=outcome_counts.values, palette="viridis")
plt.title("Match Outcomes")
plt.xlabel("Result")
plt.ylabel("Count of Matches")
plt.tight_layout()
plt.show()
```

## Section 4: Top 10 Winners

The horizontal bar chart shows the countries or teams with the most historical wins in the dataset.

```python
home_winners = df.loc[df["result"] == "Home Win", "home_team"]
away_winners = df.loc[df["result"] == "Away Win", "away_team"]
winning_teams = pd.concat([home_winners, away_winners], ignore_index=True)
top_10_winners = winning_teams.value_counts().head(10).sort_values()

plt.figure(figsize=(10, 7))
plt.barh(top_10_winners.index, top_10_winners.values, color="#264653")
plt.title("Top 10 Countries by Historical Wins")
plt.xlabel("Number of Wins")
plt.ylabel("Country or Team")
plt.tight_layout()
plt.show()
```

## Summary

These visualizations combine goal volume, match result balance, and historical winners to give a fast view of how football outcomes are distributed across the dataset.
